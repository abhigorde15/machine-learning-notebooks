```python
import matplotlib.pyplot as plt 
import numpy as np 
import pandas as pd
```

## Functional Plot

```python
x1 = np.linspace(0,5,10) #create 10 array of 10 equally spaced elements betn 0 to 5
print(x1)
y1 = x1**2
print(y1)
plt.plot(x1,y1)
plt.title("Day Squared Chart")
plt.xlabel("Days")
plt.ylabel("Day Squared") 
```

## Print Multiple Plots


```python
#plt.subplot(nrows, ncols, index)
plt.subplot(1,2,1)
plt.plot(x1,y1,'r')
plt.subplot(1,2,2)
plt.plot(x1,y1,'b')
```
