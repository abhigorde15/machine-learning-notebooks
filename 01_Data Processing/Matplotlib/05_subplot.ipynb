## Subplot()

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4]
y =[10,20,15,25]

plt.subplot(1,2,1) # 1 row, 2 cols ,1st subplot 
plt.plot(x,y)
plt.title("Line chart")

plt.subplot(1,2,2)
plt.bar(x,y)
plt.title("Bar chart")
#plt.tight_layout()

```

```python
x = [1,2,3,4]
y =[10,20,15,25]

fig,ax = plt.subplots(1,2,figsize=(10,5))
ax[0].plot(x,y)
ax[0].set_title("Line Plot")

ax[1].bar(x,y,color='green')
ax[1].set_title("Bar Chart")
plt.tight_layout()
fig.suptitle("Comparison of Line and Bar Charts")
```
