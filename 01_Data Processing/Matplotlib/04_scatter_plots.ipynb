```python
import matplotlib.pyplot as plt
```

```python
hours = [1,2,3,4,5,6,7,8]
exam_score = [50,55,60,65,70,75,80,85]
plt.scatter(hours,exam_score,color='green',marker='o',label='Student Data')
plt.xlabel('Hours Studied') 
plt.ylabel('Exam Score ')
plt.title("Relation ship between Study time and Exam score") 
plt.grid()
plt.legend()

```

```python

plt.scatter([1,2,3],[50,60,70],color='blue',label='Class A')
plt.scatter([1,2,3],[45,50,52],color='orange',label='Class B')
plt.xlabel('Hours Studied') 
plt.ylabel('Exam Score ')
plt.title("Comparison of Two classes") 
plt.grid()
plt.legend()

```
