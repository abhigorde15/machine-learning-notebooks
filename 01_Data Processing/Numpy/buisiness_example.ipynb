## Advanced Operations on Buisiness 

```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print("==== Zomato sales analysis ==== ")
print("\n Sales data shape", sales_data.shape)
print("\n Sample data for 1st 3 restau: ", sales_data[:3])
```

```python
# total sales per year 
print(np.sum(sales_data,axis=0))
print(np.sum(sales_data[:,1:],axis=0)) # here for row all rows are taken but in col range is given from 1 to last col so id will not come

#minimum sales per restaurant 
min_sales = np.min(sales_data[:,1:],axis=1)
print("minimum sales per restaurant \n",min_sales)

#maximum sales per year 
max_sales = np.max(sales_data[:,1:],axis=0)
print("Max sales per Yr \n",max_sales)

#avg sales per restaurant 
avg_sale = np.mean(sales_data[:,1:],axis=1)
print("Avg sale per rest",avg_sale)

cumsum = np.cumsum(sales_data[:,1:],axis=1)
print("Cumulative sales",cumsum)

plt.figure(figsize=(8,6))
plt.plot(np.mean(cumsum,axis=0))
plt.title("Avg Cumulative Sales across resaturant")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```
