```python
import tensorflow as tf
```

```python
x1 = tf.constant([[1,2,3],[4,5,6],[7,8,9]])
x2 = tf.constant([[1,2,3],[4,5,6],[7,8,9]])
if (x1.shape[1] == x2.shape[0]):
  mul = tf.matmul(x1,x2)
  print(mul)
else :
   print("Can't multiply")
print(x1@x2)
```

## Transpose

```python
t1_transpose = tf.transpose(x1)
print(t1_transpose)
```

```python
tensor_three1 = tf.constant([[[1,2,0],
                             [3,4,5]
                            ],
                            [[5,6,7],
                             [7,6,5]
                             ],
                            [[2,3,4],
                             [5,6,7]
                            ]
                            ],dtype=tf.float32)
tensor_three2 = tf.constant([[[1,2,0],
                             [3,4,5],
                              [3,4,5]],
                            [[5,6,7],
                             [7,6,5],
                             [3,4,5]],
                            [[2,3,4],
                             [5,6,7],
                            [3,4,5]]
                            ],dtype=tf.float32)
print(tf.linalg.matmul(tensor_three1,tensor_three2))  #each 2d tensor * with each other at similar indices
```

##Sparse Tensor

```python
sparse_tensor = tf.sparse.SparseTensor(indices=[[1,2],[3,3]],values=[1,2],dense_shape=[5,5])
print(sparse_tensor)
```

Band Part

```python
t1 = tf.constant([[1,2,3],
             [-2,-3,5],
             [-2,1,4]])
band_part = tf.linalg.band_part(t1,0,2)
print(band_part)
```

```python
t1 = tf.constant([[1,2,3],
             [-2,-3,5],
             [-2,1,4]])
band_part = tf.linalg.band_part(t1,1,1)
print(band_part)
```

## Inverse of Matrix

```python
A = tf.constant([[1.0, 2.0],
                 [3.0, 4.0]])

inv = tf.linalg.inv(A)
print(inv)
print(A@inv)##identity matrix
```

## SVD of matrix

```python
A = tf.constant([[1, 2,3],
                 [3, 4,5],
                 [6, 7,8]
                 ],dtype=tf.float32)
s,v,d = tf.linalg.svd(A)
print(s)
print(v)
print(d)
```

##EinSum Operator

```python
A = tf.constant([1,2,3])
B = tf.constant([4,5,6])
tf.einsum('i,j->ij',A,B) # element wise dot product
```

```python
#2-D
A = tf.constant([[1,2],[3,4]])
B = tf.constant([[5,6],[7,8]])
tf.einsum('ij,jk->ik',A,B) # same as matmul(A,B)
```

```python
tf.einsum("ij->ji",A) # same as tf.transpose(A)
```

```python
A = tf.constant([[1, 2], [3, 4]])
B = tf.constant([[10, 20], [30, 40]])
tf.einsum("ij,ij->ij", A, B)  # same as tf.multiply(A, B)
```
