```python
import tensorflow as tf
```

## Basics

```python
tensor_zerod = tf.constant(4)
print(tensor_zerod)
```

```python
tensor_one = tf.constant([2.0, 4.0, 6.0])
print(tensor_one)
```

```python
tensor_two = tf.constant([[1, 2], [3, 4]])
print(tensor_two)
```

```python
tensor_three = tf.constant([[[1,2,0],
                             [3,4,5]],
                            [[5,6,7],
                             [7,6,5]],
                            [[2,3,4],
                             [5,6,7]]
                            ],dtype=tf.float32)

print(tensor_three)
```

```python
print(tensor_one.ndim)
print(tensor_three.ndim)
```

# Converting float to int

```python
#tf_float = tf.constant([1,2,3,4.5],dtype=tf.int32)
tf_float = tf.cast([1,2,3,4.5],dtype=tf.int32)
print (tf_float)
```

## using Bool tensor


```python
casted_tensor = tf.cast(tensor_three,dtype=tf.bool)
print(casted_tensor) # only zero become false
```

## Tensor String

```python
#tensor_str = tf.constant(["Hello",1,"World"]);
tensor_str = tf.constant(["Hello","World"]);
print(tensor_str)
```

## Converting numpy array to tensor

```python
import numpy as np
np_array = np.array([1,3,4,5])
print(np_array)
converted_arr = tf.convert_to_tensor(np_array)
print(converted_arr)
```
