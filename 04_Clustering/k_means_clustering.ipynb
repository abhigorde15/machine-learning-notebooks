# K-Means Clustering

## Importing the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importing the dataset

```python
dataset = pd.read_csv('Mall_Customers.csv')
X = dataset.iloc[:,[3,4]].values 
```

## Using the elbow method to find the optimal number of clusters

```python
from sklearn.cluster import KMeans 
wcss =[]
for i in range(1,11):
    kmean = KMeans(n_clusters=i,init='k-means++',random_state=42)
    kmean.fit(X)
    wcss.append(kmean.inertia_)
plt.plot(range(1,11),wcss)
plt.title("Elbow Method")
plt.xlabel("No. of Clusters")
plt.ylabel("WCSS")
plt.show() ## from graph 5 is answer


```

## Training the K-Means model on the dataset

```python
kmean = KMeans(n_clusters=5,init='k-means++',random_state=42)
y_kmean =  kmean.fit_predict(X)
```

## Visualising the clusters

```python
plt.scatter(X[y_kmean == 0,0],X[y_kmean == 0,1],s=100,c='red', label='Cluster1')
plt.scatter(X[y_kmean == 1,0],X[y_kmean == 1,1],s=100,c='blue', label='Cluster2')
plt.scatter(X[y_kmean == 2,0],X[y_kmean == 2,1],s=100,c='green', label='Cluster3')
plt.scatter(X[y_kmean == 3,0],X[y_kmean == 3,1],s=100,c='yellow', label='Cluster4')
plt.scatter(X[y_kmean == 4,0],X[y_kmean == 4,1],s=100,c='magenta', label='Cluster1')
plt.scatter(kmean.cluster_centers_[:,0],kmean.cluster_centers_[:,1],s=300,c='black',label='centroid') 
plt.title("K-mean Cluster")
plt.xlabel("Anual Income")
plt.ylabel("Spending Score(1-100)")
plt.legend()
plt.show()
```
